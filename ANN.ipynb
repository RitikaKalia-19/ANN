**3.	Implement a neuron with 5 inputs.**

```python
import torch

x = torch.tensor([2., 3., 4., 5., 6.])
w = torch.tensor([0.5, 0.8, 0.2, 0.3,0.7])
b = torch.tensor(1.)

z = torch.dot(x, w) + b
print("Weighted Sum =", z)
```

**4.	Change the weights and bias; predict the output before execution**

**1. The Dot Product Calculation `torch.dot(x, w)`:**
* 2.0 * 0.10 = 0.2
* 3.0 * 0.20 = 0.6
* 4.0 * 0.40 = 1.6
* 5.0 * 0.10 = 0.5
* 6.0 * 0.20 = 1.2

* **Total Sum of Products:** 0.2 + 0.6 + 1.6 + 0.5 + 1.2 = 4.1

**2. Adding the Bias (`+ b`):**
* 4.1 + (-1.0) = **3.1**

```python
import torch

x = torch.tensor([2., 3., 4., 5., 6.])
w = torch.tensor([0.10, 0.2, 0.4, 0.1,0.2])
b = torch.tensor(-1.)

z = torch.dot(x, w) + b
print("Weighted Sum =", z)
```

```python
print(torch.sigmoid(z))
print(torch.tanh(z))
print(torch.relu(z))
```

**5.	Plot Sigmoid, Tanh and ReLU.**

```python
import matplotlib.pyplot as plt
x=torch.linspace(-10,10,200)
plt.plot(x,torch.sigmoid(x), label='sigmoid')
plt.plot(x,torch.tanh(x), label='tanh')
plt.plot(x,torch.relu(x), label='ReLU')
plt.grid()
plt.legend()
plt.show()
```

**6.	Use Autograd to compute the gradient of y = x² + 5x + 2.**

```python
x=torch.tensor(2.0, requires_grad=True)
y=x**2+3*x+1
y.backward()
print(x.grad)
```

**7.	Create nn.Linear(5, 1) and inspect its weights.**

```python
import torch.nn as nn
layer=nn.Linear(5,1)
print(layer.weight)
print(layer.bias)
inp=torch.tensor([2.,3.,4.,5.,8.])
print(layer(inp))
```

**8.	Compare the manual neuron implementation with nn.Linear.**


| Feature | Manual Implementation (`torch.dot`) | PyTorch Layer (`nn.Linear`) |
| :--- | :--- | :--- |
| **Weights & Bias** | You must manually create and define every single number. | PyTorch automatically creates them using random numbers. |
| **Scaling Up** | Hard. If you want 100 neurons, you have to write 100 dot products manually. | Easy. You just change the numbers inside the function (e.g., `nn.Linear(5, 100)`). |
| **AI Training** | You have to manually tell PyTorch to track gradients (`requires_grad=True`). | PyTorch automatically tracks gradients so the weights can learn. |
